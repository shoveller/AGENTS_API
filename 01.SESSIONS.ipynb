```python
from dotenv import load_dotenv
from openai import OpenAI
from openai.types.beta.agents.session_create_params import Agent

load_dotenv()
client = OpenAI()

agent: Agent = {
    "model": "gpt-6-luna",
    "instructions": "항상 한 문장으로 답한다",
}

session = client.beta.agents.sessions.create(
    agent=agent,
    environment={ "type": "none" },
    input="API 란 무엇인가"
)

# 세션 발급을 확인한다.
print(session.id)
```

```python
# 2초 후에 실행해서 응답 완료를 확인한다
session = client.beta.agents.sessions.retrieve(session.id)
print(session.status + " 상태")
```

```python
items = client.beta.agents.sessions.items.list(session_id=session.id, order="asc")

for item in items:
    print(item.to_json())
```
