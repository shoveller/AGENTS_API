```python
from dotenv import load_dotenv
from openai import OpenAI
from openai.types.beta.agents.session_create_params import Agent

load_dotenv()
client = OpenAI()

model = "gpt-6-luna"    

# 에이전트를 생성한다.
agent = client.beta.agents.create(
    model=model,
    instructions="항상 왕의 말투로 답한다",
    reasoning={
        "effort": "low"
    },
    text={
        # 응답 상세 정도 조절
        "verbosity": "low"
    }
)

print(agent.id)
```

```python
# 세션을 생성한다
session = client.beta.agents.sessions.create(
    agent_id=agent.id,
    environment={
        "type": "none"
    },
    input="파이썬의 dict 자료형을 설명"
)
print(session.id)
```

```python
# 에이전트 정보를 업데이트한다.
agent = client.beta.agents.update(agent.id, instructions="항상 왕의 말투로 답한다. 그리고 항상 존댓말을 사용한다.")
print(agent.id)
```

```python
# 기존 세션에 추가 메시지를 보낸다. 반환값은 없다.
client.beta.agents.sessions.events.create(
    session_id=session.id,
    events=[{
        "type":"agent.session.input.message",
        "input":[{
            "role": "user",
            "content": [
                {
                    "type": "input_text",
                    "text": "비동기를 설명"
                }
            ]
        }]
    }]
)
```

```python
# 대화를 출력한다.
items = client.beta.agents.sessions.items.list(session_id=session.id, order="asc")

for item in items:
    print(item.content[0].text)
    print("=====================================")
```

```python
# 새로운 세션을 생성한다
new_session = client.beta.agents.sessions.create(
    agent_id=agent.id,
    environment={
        "type": "none"
    },
    input="수원 왕갈비 레시피를 2줄로 설명"
)
print(new_session.id)
```

```python
items = client.beta.agents.sessions.items.list(session_id=new_session.id, order="asc")

for item in items:
    print(item.content[0].text)
    print("=====================================")
```
